```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Define Functions and Derivatives
# Values and formulas derived from Task 2 and manual calculations
functions = {
    'f1': {'func': lambda x: x**2 - 4, 'deriv': lambda x: 2*x},
    'f2': {'func': lambda x: x**3 - x - 2, 'deriv': lambda x: 3*x**2 - 1},
    'f3': {'func': lambda x: np.exp(x) - 3, 'deriv': lambda x: np.exp(x)},
    'f4': {'func': lambda x: x**3 - 6*x**2 + 11*x - 6, 'deriv': lambda x: 3*x**2 - 12*x + 11}
}
```

```python
# 2. Algorithm Implementations with Stopping Criteria
def run_bisection(f_key, a, b, eps=1e-6):
    f = functions[f_key]['func']
    data = []
    for k in range(1, 5): # 4 iterations as per manual task
        c = (a + b) / 2
        fc = f(c)
        data.append({'k': k, 'a': a, 'b': b, 'x_k': c, 'f(x_k)': fc, 'error': abs(b-a)/2})
        if abs(fc) < eps or (abs(b-a)/2) < eps: break
        if f(a) * fc < 0: b = c
        else: a = c
    return pd.DataFrame(data)

def run_newton(f_key, x0, eps=1e-6):
    f = functions[f_key]['func']
    df = functions[f_key]['deriv']
    data = []
    xk = x0
    for k in range(4):
        fk, dfk = f(xk), df(xk)
        corr = fk / dfk
        x_next = xk - corr
        err = abs(x_next - xk)
        data.append({'k': k, 'x_k': xk, 'f(x_k)': fk, 'f_prime': dfk, 'corr': corr, 'x_next': x_next, 'error': err})
        if abs(fk) < eps or err < eps: break
        xk = x_next
    return pd.DataFrame(data)

def run_secant(f_key, x_prev, x_curr, eps=1e-6):
    f = functions[f_key]['func']
    data = []
    for k in range(1, 5):
        f_prev, f_curr = f(x_prev), f(x_curr)
        x_next = x_curr - f_curr * (x_curr - x_prev) / (f_curr - f_prev)
        err = abs(x_next - x_curr)
        data.append({'k': k, 'x_k-1': x_prev, 'x_k': x_curr, 'f(x_curr)': f_curr, 'x_next': x_next, 'error': err})
        if abs(f_curr) < eps or err < eps: break
        x_prev, x_curr = x_curr, x_next
    return pd.DataFrame(data)
```

```python
# 3. Execution for all functions based on Task 2 parameters
# Data maps to "Screenshot 1447-11-22 at 2.02.28 AM.png"
results = {
    'f1_Bisection': run_bisection('f1', 1.0, 3.0),
    'f1_Newton': run_newton('f1', 1.5),
    'f1_Secant': run_secant('f1', 1.0, 3.0), # From "Screenshot 1447-11-22 at 2.02.41 AM.png"
    'f2_Bisection': run_bisection('f2', 1.0, 2.0),
    'f2_Newton': run_newton('f2', 1.0),
    'f3_Newton': run_newton('f3', 1.0),
    'f4_Bisection': run_bisection('f4', 2.5, 3.5)
}
```
